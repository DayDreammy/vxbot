```python
import sqlite3
import pandas as pd
from pathlib import Path
from datetime import datetime, timedelta

def get_articles_by_title_keyword(keyword, db_path='data/message_monitor.db'):
    """
    通过标题关键词查询文章
    
    Parameters:
    -----------
    keyword : str
        要搜索的关键词
    db_path : str
        数据库路径
        
    Returns:
    --------
    pd.DataFrame
        包含查询结果的DataFrame
    """
    query = """
    SELECT 
        id,
        message_id,
        from_user,
        title,
        url,
        content,
        summary,
        created_at,
        processed,
        process_time,
        pdf_path
    FROM wechat_articles
    WHERE title LIKE ?
    ORDER BY created_at DESC
    """
    
    with sqlite3.connect(db_path) as conn:
        # 使用 LIKE 进行模糊匹配，%表示任意字符
        df = pd.read_sql_query(query, conn, params=[f'%{keyword}%'])
    
    # 转换时间列
    df['created_at'] = pd.to_datetime(df['created_at'])
    df['process_time'] = pd.to_datetime(df['process_time'])
    
    return df
```

```python

# 使用示例
if __name__ == "__main__":
    # 搜索标题中包含"芯片"的文章
    keyword = "广州市番禺区小谷围街道办事处招聘工作人员公告"
    df = get_articles_by_title_keyword(keyword)
    
    print(f"找到包含关键词 '{keyword}' 的文章数量：{len(df)}")
    print("\n文章列表：")
    
    # 显示查询结果
    for _, row in df.iterrows():
        print(f"\n标题: {row['title']}")
        print(f"来源: {row['from_user']}")
        print(f"发布时间: {row['created_at']}")
        print(f"URL: {row['url']}")
        try:    
            print(f"内容: {row['content']}")
        except:
            print(f"内容: ")
        print("-" * 80)
```
